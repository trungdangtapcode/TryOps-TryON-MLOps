# VTON Baseline Exploration

This notebook is intentionally exploratory. Production execution uses `scripts/run_vton_baseline.py` and the pipeline modules under `src/tryops/pipelines/`.

Purpose:

- Create deterministic synthetic inputs.
- Run the naive VTON baseline.
- Compare two baseline configurations.
- Inspect generated artifacts and failure labels.


```python
from pathlib import Path
from tryops.pipelines.vton_comparison import compare_vton_baselines
from tryops.simple_image import RgbImage, solid_rgb, write_png_rgb
```

Create small synthetic person and garment images. Real VTON evaluation should use licensed public datasets only after dataset cards and checksums are complete.

```python
root = Path('../artifacts/notebooks/vton_baseline')
root.mkdir(parents=True, exist_ok=True)
person = root / 'person.png'
garment = root / 'garment.png'
write_png_rgb(person, solid_rgb(180, 240, (180, 190, 205)))
write_png_rgb(garment, solid_rgb(96, 96, (40, 80, 190)))
```

Run the comparison. This is a proxy-quality baseline only; the target model remains CatVTON, with IDM-VTON as a stretch target.

```python
comparison = compare_vton_baselines(
    person_image_path=person,
    garment_image_path=garment,
    output_dir=root / 'comparison',
    cache_dir=root / 'cache',
)
comparison['winner_by_structural_similarity']
```
